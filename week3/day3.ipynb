```python
import os
import json
import requests
from dotenv import load_dotenv, find_dotenv
from huggingface_hub import login
from transformers import AutoTokenizer
```

```python
env_path = find_dotenv(".env", usecwd=True)
load_dotenv(env_path, override=True)
HF_TOKEN = (os.getenv("HUGGINGFACE_HUB_TOKEN"))
```

```python
HF_API_URL = os.getenv("HF_API_URL", "https://api-inference.huggingface.co")
os.environ.setdefault("HUGGINGFACE_HUB_TOKEN", HF_TOKEN)
os.environ.setdefault("HF_TOKEN", HF_TOKEN)
print(HF_TOKEN)
```

```python
tokenizer = AutoTokenizer.from_pretrained(
    "meta-llama/Meta-Llama-3.1-8B",
    token=HF_TOKEN,          # <- imprescindible para modelos con licencia
    trust_remote_code=True
)
```

```python
text = "Estoy entusiasmado por mostrarles a mis ingenieros de LLM los Tokenizadores en acción."
tokens = tokenizer.encode(text)
tokens
```

```python
len(tokens)
```

```python
tokenizer.decode(tokens)
```

```python
tokenizer.batch_decode(tokens)
```

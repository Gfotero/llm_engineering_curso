```python
import torch

def check_pytorch_cuda():
    """
    Verifica la instalación de PyTorch, la disponibilidad de CUDA 
    y muestra información sobre la GPU.
    """
    print(f"PyTorch Version: {torch.__version__}")
    
    # Verificar si CUDA está disponible
    cuda_disponible = torch.cuda.is_available()
    print(f"CUDA disponible: {cuda_disponible}")
    
    if cuda_disponible:
        # Contar el número de GPUs
        num_gpus = torch.cuda.device_count()
        print(f"Número de GPUs disponibles: {num_gpus}")
        
        # Obtener el nombre de la GPU actual (generalmente la 0)
        gpu_name = torch.cuda.get_device_name(0)
        print(f"Nombre de la GPU: {gpu_name}")
    else:
        print("PyTorch no puede encontrar una GPU compatible con CUDA. 😟")
        print("Se ejecutará en la CPU.")

if __name__ == "__main__":
    check_pytorch_cuda()
```
